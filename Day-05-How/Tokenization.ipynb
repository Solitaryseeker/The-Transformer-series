```
import torch
import torch.nn as nn
```

```
text = "the cat chased the mouse"
print(text)
```

```
tokens = text.split()
print("Tokens:",tokens)
```

```
vocab = {token: idx for idx, token in enumerate(sorted(set(tokens)))}
print("Vocabulary: ",vocab)
```

```
token_ids = [vocab[token] for token in tokens]
print("Token IDs:",token_ids)
```

```
token_ids = [vocab[token] for token in tokens]
print("Token IDs:",token_ids)

# Convert to PyTorch tensor
token_ids = torch.tensor(token_ids, dtype=torch.long)
```

```
embedding_dim = 8
embedding_layer = nn.Embedding(num_embeddings=len(vocab),embedding_dim=embedding_dim)
embeddings = embedding_layer(token_ids)
print("Embedding Shape:",embeddings.shape)
print("Embeddings:",embeddings)
```

```
print("TEXT → TOKENS → TOKEN IDs → EMBEDDINGS")

print("Text:",text)

print("Tokens:",tokens)

print("Token IDs:",token_ids.tolist())

print("Embedding Shape:",embeddings.shape)

print(embeddings)
```
