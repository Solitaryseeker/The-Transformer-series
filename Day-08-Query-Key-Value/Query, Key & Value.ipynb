```python
import torch
import torch.nn as nn

# Example sentence
tokens = ["The", "cat", "chased", "the", "mouse"]

# Simple token IDs for demonstration
token_ids = torch.tensor([0, 1, 2, 3, 4])

# Dimensions
vocab_size = 5
embedding_dim = 8
attention_dim = 4

# Token embedding layer
embedding = nn.Embedding(vocab_size, embedding_dim)

# Convert token IDs into embeddings
x = embedding(token_ids)

print("Input shape:", x.shape)
```

```python
W_Q = nn.Linear(embedding_dim, attention_dim)
W_K = nn.Linear(embedding_dim, attention_dim)
W_V = nn.Linear(embedding_dim, attention_dim)

Q = W_Q(x)
K = W_K(x)
V = W_V(x)

print("Q shape:", Q.shape)
print("K shape:", K.shape)
print("V shape:", V.shape)
```

```python
print("\nQuery:")
print(Q)

print("\nKey:")
print(K)

print("\nValue:")
print(V)
```
